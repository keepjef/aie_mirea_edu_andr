# Основы работы с массивами в библиотеке NumPy #
В этом ноутбуке представлены примеры создания, манипуляции и выполнения математических операций над массивами.

```python
#1. Подключение библиотеки
import numpy as np
```

```python
#2. Генерация массивов
arr = np.array([10, 20, 30, 40, 50])  
print(f"\nОдномерный массив (вектор):\n{arr}")

arr_d = np.array([[5, 10, 15], [20, 25, 30]])
print(f"\nДвумерный массив (матрица 2x3):\n{arr_d}")

arr_zeros = np.zeros(8, dtype=int)
print(f"\nВектор из восьми нулей:\n{arr_zeros}")

arr_zeros_d = np.zeros((3, 4))
print(f"\nМатрица нулей размером 3x4:\n{arr_zeros_d}")

arr_ones = np.ones(4, dtype=int)
print(f"\nВектор из единиц:\n{arr_ones}")

arr_ones_d = np.ones((2, 2))
print(f"\nКвадратная матрица из единиц (2x2):\n{arr_ones_d}")

arr_arange = np.arange(5, 11)
print(f"\nДиапазон чисел от 5 до 10:\n{arr_arange}")

arr_step = np.arange(0, 50, 5)
print(f"\nПоследовательность от 0 до 50 с шагом 5:\n{arr_step}")

arr_random = np.random.random((3, 2))
print(f"\nМатрица случайных чисел (3x2):\n{arr_random}")

arr_1d = np.arange(20)
print(f"\nИсходный вектор (0-19):\n {arr_1d}")
# Изменение формы массива на 4 строки и 5 столбцов
arr_2d = arr_1d.reshape(4, 5)
print(f"\nПреобразованный массив 4x5:\n{arr_2d}")
```

```python
#3. Просмотр свойств массива
print("Параметры массива arr_zeros_d:")
print(f"Размерность (shape): {arr_zeros_d.shape}")
print(f"Тип данных (dtype): {arr_zeros_d.dtype}")
```

```python
#4. Доступ к элементам и срезы
print(f"Пятый элемент массива arr_step (индекс 4): {arr_step[4]}")
print(f"Выборка элементов со 2-го по 5-й: {arr_step[2:6]}")

# Работа с многомерным массивом (arr_2d - 4x5)
print(f"\nЗначение в 3-й строке и 4-м столбце: {arr_2d[2, 3]}") 
print(f"\nПервые две строки массива:\n{arr_2d[:2]}")
print(f"\nСтолбцы с индексами 2 и 3 для всех строк:\n{arr_2d[:, 2:4]}")
print(f"\nПодмассив: строки 1-2 и столбцы 3-4:\n{arr_2d[1:3, 3:5]}")
```

```python
#5. Арифметика с массивами и числами (Broadcasting)
arr_calc = np.arange(1, 6) # [1, 2, 3, 4, 5]
num = 10
print(f"Исходный массив: {arr_calc}")
print(f"Прибавили {num}: {arr_calc + num}")
print(f"Отняли {num}: {arr_calc - num}")
print(f"Умножили на {num}: {arr_calc * num}")
print(f"Разделили на {num}: {arr_calc / num}")
print(f"Возвели в степень {num}: {arr_calc ** num}")
```

```python
# Операции между двумя матрицами
X = np.array([[10, 20], [30, 40]])
Y = np.array([[2, 4], [5, 8]])
print("Матрица X:\n", X)
print("Матрица Y:\n", Y)

print(f"Сумма (X + Y):\n{X + Y}")
print(f"Разность (X - Y):\n{X - Y}")
print(f"Произведение элементов (X * Y):\n{X * Y}")
print(f"Деление элементов (X / Y):\n{X / Y}")
print(f"Возведение X в степень Y:\n{X ** Y}\n")
```

```python
# Агрегирующие функции (статистика)
print(f"Среднее арифметическое X: {X.mean()}")
print(f"Сумма всех значений X: {X.sum()}")
print(f"Стандартное отклонение X: {X.std():.3f}") 

# Вычисления вдоль осей
print(f"Средние значения по строкам (axis=1): {X.mean(axis=1)}")
print(f"Средние значения по столбцам (axis=0): {X.mean(axis=0)}")
```
